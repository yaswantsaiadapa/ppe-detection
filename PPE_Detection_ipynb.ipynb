```python
from PIL import Image
import matplotlib.pyplot as plt
import os
import random

image_folder = "PPE-detection-1/train/images"

images = random.sample(os.listdir(image_folder), 4)

plt.figure(figsize=(12,8))

for i,img in enumerate(images):
    plt.subplot(2,2,i+1)
    image = Image.open(os.path.join(image_folder,img))
    plt.imshow(image)
    plt.axis("off")

plt.show()
```
